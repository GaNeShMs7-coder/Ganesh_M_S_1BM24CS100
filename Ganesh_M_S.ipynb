```python
print("Ganesh Muralidhar Sureban")
print("1BM24CS100")

print("___________________________")
print("")
import random

def print_board(board):
    """Prints the current state of the 3x3 game board."""
    print(f"\n {board[0]} | {board[1]} | {board[2]} ")
    print("---|---|---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---|---|---")
    print(f" {board[6]} | {board[7]} | {board[8]} \n")

def check_winner(board, player):
    """Checks if the given player ('X' or 'O') has won the game."""
    win_conditions = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],  # Rows
        [0, 3, 6], [1, 4, 7], [2, 5, 8],  # Columns
        [0, 4, 8], [2, 4, 6]              # Diagonals
    ]
    for condition in win_conditions:
        if board[condition[0]] == board[condition[1]] == board[condition[2]] == player:
            return True
    return False

def is_board_full(board):
    """Returns True if there are no empty spaces left on the board."""
    return "-" not in board

def get_human_move(board):
    """Prompts the human player to enter a valid move (1-9)."""
    while True:
        try:
            move = int(input("Enter your move (1-9): ")) - 1
            if move in range(0, 9) and board[move] == "-":
                return move
            else:
                print("That spot is taken or invalid. Please try again.")
        except ValueError:
            print("Invalid input. Please enter a number between 1 and 9.")

def get_computer_move(board):
    """Generates a random available move for the PC player."""
    available_moves = [i for i, spot in enumerate(board) if spot == "-"]
    return random.choice(available_moves)

def play_game():
    """Main game loop for Tic-Tac-Toe."""
    # Initialize a clean board with empty slot placeholders
    board = ["-"] * 9

    print("Welcome to Tic-Tac-Toe!")
    print("You are 'X' and the Computer is 'O'. Positions are numbered 1 to 9.")

    # Helper reference guide for grid positions
    print("\nPosition Guide:")
    print(" 1 | 2 | 3 \n---|---|---\n 4 | 5 | 6 \n---|---|---\n 7 | 8 | 9 ")

    current_player = "X"  # Human starts first

    while True:
        print_board(board)

        if current_player == "X":
            print("Your turn...")
            move = get_human_move(board)
            board[move] = "X"

            if check_winner(board, "X"):
                print_board(board)
                print(" Congratulations! You beat the computer!")
                break
            current_player = "O"
        else:
            print("Computer is making a move...")
            move = get_computer_move(board)
            board[move] = "O"
            print(f"Computer chose position {move + 1}")

            if check_winner(board, "O"):
                print_board(board)
                print("Game over! The computer wins!")
                break
            current_player = "X"

        if is_board_full(board):
            print_board(board)
            print(" It's a draw!")
            break

if __name__ == "__main__":
    play_game()

```

```python
def reflex_agent(location, status):
    if status == "Dirty":
        return "Suck"
    elif location == "A":
        return "Right"
    elif location == "B":
        return "Left"

def run_environment():
    environment = {
        "A": input("Is room A 'Dirty' or 'Clean'? ").strip().capitalize(),
        "B": input("Is room B 'Dirty' or 'Clean'? ").strip().capitalize()
    }

    current_location = input("Start location ('A' for Left, 'B' for Right): ").strip().upper()

    if current_location not in ["A", "B"] or environment["A"] not in ["Clean", "Dirty"] or environment["B"] not in ["Clean", "Dirty"]:
        print("Invalid inputs provided.")
        return

    total_cost = 0
    print("\n--- Starting Simulation ---")

    while environment["A"] == "Dirty" or environment["B"] == "Dirty":
        status = environment[current_location]
        action = reflex_agent(current_location, status)

        print(f"Location: {current_location} | Status: {status} | Action: {action}")

        if action == "Suck":
            environment[current_location] = "Clean"
            total_cost += 1
        elif action == "Right":
            current_location = "B"
            total_cost += 1
        elif action == "Left":
            current_location = "A"
            total_cost += 1

    print("\nAll rooms are Clean!")
    print(f"Total Path Cost / Energy Expended: {total_cost}")

run_environment()
```
