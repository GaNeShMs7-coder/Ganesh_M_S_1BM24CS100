```python
print("GANESH M S")
print("1BM24CS100")
print("AI_LAB-H_WEEK02")
# 8-Puzzle Problem Solver using Depth-First Search (DFS)
import sys

# Increase recursion depth limit just in case for deep DFS branches
sys.setrecursionlimit(100000)

class PuzzleState:
    def __init__(self, board, parent=None, move="", depth=0):
        self.board = board       # 1D tuple representing the 3x3 board
        self.parent = parent     # Pointer to the parent node to reconstruct the path
        self.move = move         # The move that led to this state (Up, Down, Left, Right)
        self.depth = depth       # Current depth level in the search tree

    def __eq__(self, other):
        return self.board == other.board

    def __hash__(self):
        return hash(self.board)

def get_possible_moves(state):
    """Generates all valid next states by moving the blank tile (0)."""
    moves = []
    board = list(state.board)
    blank_idx = board.index(0)
    row, col = blank_idx // 3, blank_idx % 3

    # Define directional offsets: (row_change, col_change, move_name)
    directions = [(-1, 0, "Up"), (1, 0, "Down"), (0, -1, "Left"), (0, 1, "Right")]

    for dr, dc, move_name in directions:
        new_row, new_col = row + dr, col + dc
        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_idx = new_row * 3 + new_col
            # Clone and swap tiles
            new_board = board[:]
            new_board[blank_idx], new_board[new_idx] = new_board[new_idx], new_board[blank_idx]
            moves.append(PuzzleState(tuple(new_board), state, move_name, state.depth + 1))
    return moves

def solve_dfs(initial_board, goal_board, depth_limit=50):
    """Solves the 8-puzzle using Depth-Limited DFS to handle infinite loops or deep paths."""
    start_state = PuzzleState(tuple(initial_board))
    goal_state = PuzzleState(tuple(goal_board))

    # Using an explicit Stack for DFS: stores tuples of (state)
    stack = [start_state]
    # Visited tracking to avoid cyclical paths
    visited = set()

    nodes_expanded = 0

    while stack:
        current_state = stack.pop()
        nodes_expanded += 1

        if current_state.board == goal_state.board:
            # Reconstruct path
            path = []
            curr = current_state
            while curr.parent:
                path.append((curr.move, curr.board))
                curr = curr.parent
            path.reverse()
            return path, nodes_expanded

        if current_state not in visited and current_state.depth < depth_limit:
            visited.add(current_state)
            # Add neighbors to stack (DFS behavior)
            for neighbor in get_possible_moves(current_state):
                if neighbor not in visited:
                    stack.append(neighbor)

    return None, nodes_expanded

def print_board(board_tuple):
    for i in range(0, 9, 3):
        print(f" {board_tuple[i]} {board_tuple[i+1]} {board_tuple[i+2]} ")
    print("-" * 11)

# --- Interactive Input and Execution Setup ---
print("--- Enter the 8-Puzzle Initial State (use 0 for blank space) ---")
print("Enter elements row by row separated by spaces (e.g., 1 2 3):")
initial = []
for i in range(3):
    row = list(map(int, input(f"Row {i+1}: ").split()))
    initial.extend(row)

print("\n--- Enter the 8-Puzzle Goal State ---")
goal = []
for i in range(3):
    row = list(map(int, input(f"Row {i+1}: ").split()))
    goal.extend(row)

print("\nInitial State:")
print_board(initial)
print("Goal State:")
print_board(goal)

print("\nSolving using DFS... Please wait...")
# Depth limit safety to prevent colab crashes since pure DFS can wander infinitely
solution_path, total_nodes = solve_dfs(initial, goal, depth_limit=30)

if solution_path is not None:
    print(f"\nGoal Reached in {len(solution_path)} steps!")
    print(f"Total nodes processed: {total_nodes}\n")
    print("Step-by-Step Path:")
    print_board(initial)
    for move, board in solution_path:
        print(f"Move: {move}")
        print_board(board)
else:
    print(f"\n❌ No solution found within depth limit or puzzle is unsolvable.")
    print(f"Total nodes evaluated: {total_nodes}")
```
